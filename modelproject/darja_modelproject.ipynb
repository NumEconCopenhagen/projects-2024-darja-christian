```python
import sympy as sm
from scipy import optimize
from darja_modelproject import stackelbergduopolClass
model = stackelbergduopolClass()
```

```python
## inverse demand function
def P(x_1, x_2, a = 5, b = 1/4): ## local scope 
    c = a * 1/b
    if 0 <= (x_1 + x_2) <= c: 
        result =  a - b*(x_1 + x_2)
    else: ## can't have a negative price
        result = 0
    return result
```

```python
## solution using sympy 

x_1 = sm.symbols("x_1")
x_2 = sm.symbols("x_2")
#a = sm.symbols("a")
a = 5
#b = sm.symbols("b")
b = 1/4
#p = sm.symbols("p")
p_1 = 2
p_2 = 1
```

```python
inverse_demand =  a-b*(x_1 + x_2)
inverse_demand
```

```python
objective_1 = inverse_demand * x_1 - p_1*x_1
objective_1
```

```python
objective_2 = inverse_demand * x_2 - p_2*x_2
objective_2
```

```python
foc = sm.diff(objective_2, x_2)
foc
```

```python
sol = sm.solve(sm.Eq(foc,0), x_2)
sol ## best answer function of firm 2 
```

```python
sub_objective_1= objective_1.subs(x_2, sol[0])
sub_objective_1
```
