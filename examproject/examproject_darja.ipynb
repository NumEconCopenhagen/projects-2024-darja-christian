## 1. <a id='toc1_'></a>[Problem 1: Production economy and CO2 taxation](#toc0_)

Consider a production economy with two firms indexed by $j \in \{1,2\}$. Each produce its own good. They solve

$$
\begin{align*}
\max_{y_{j}}\pi_{j}&=p_{j}y_{j}-w_{j}\ell_{j}\\\text{s.t.}\;&y_{j}=A\ell_{j}^{\gamma}.
\end{align*}
$$

Optimal firm behavior is

$$
\begin{align*}
\ell_{j}^{\star}(w,p_{j})&=\left(\frac{p_{j}A\gamma}{w}\right)^{\frac{1}{1-\gamma}} \\
y_{j}^{\star}(w,p_{j})&=A\left(\ell_{j}^{\star}(w,p_{j})\right)^{\gamma}
\end{align*}
$$

The implied profits are

$$
\pi_{j}^*(w,p_{j})=\frac{1-\gamma}{\gamma}w\cdot\left(\frac{p_{j}A\gamma}{w}\right)^{\frac{1}{1-\gamma}}
$$

A single consumer supplies labor, and consumes the goods the firms produce. She also recieves the implied profits of the firm.<br>
She solves:

$$
\begin{align*}
U(p_1,p_2,w,\tau,T) = \max_{c_{1},c_{2},\ell} & \log(c_{1}^{\alpha}c_{2}^{1-\alpha})-\nu\frac{\ell^{1+\epsilon}}{1+\epsilon} \\
\text{s.t.}\,\,\,&p_{1}c_{1}+(p_{2}+\tau)c_{2}=w\ell+T+\pi_{1}^*(w,p_{1})+\pi_{2}^*(w,p_{2})
\end{align*}
$$

where $\tau$ is a tax and $T$ is lump-sum transfer. <br>
For a given $\ell$, it can be shown that optimal behavior is

$$
\begin{align*}
c_{1}(\ell)&=\alpha\frac{w\ell+T+\pi_{1}^*(w,p_{1})+\pi_{2}^*(w,p_{2})}{p_{1}} \\
c_{2}(\ell)&=(1-\alpha)\frac{w\ell+T+\pi_{1}^*(w,p_{1})+\pi_{2}^*(w,p_{2})}{p_{2}+\tau} \\
\end{align*}
$$
Such that optimal behavior is:
$$
\ell^* = \underset{\ell}{\arg\max} \log(\left(c_{1}(\ell)\right)^{\alpha}\cdot \left(c_{2}(\ell)\right)^{1-\alpha})-\nu\frac{\ell^{1+\epsilon}}{1+\epsilon} 
$$
With optimal consumption:
$$
\begin{align*}
c_1^*=c_{1}(\ell^*) \\
c_2^*=c_{2}(\ell^*)\\
\end{align*}
$$


The government chooses $\tau$ and balances its budget so $T=\tau c_2^*$. We initially set $\tau,T=0$.

Market clearing requires:

1. Labor market: $\ell^* = \ell_1^* + \ell_2^*$
1. Good market 1: $c_1^* = y_1^*$
1. Good market 2: $c_2^* = y_2^*$

## 2. <a id='toc2_'></a>[Problem 2: Career choice model](#toc0_)

Consider a graduate $i$ making a choice between entering $J$ different career tracks. <br>
Entering career $j$ yields utility $u^k_{ij}$. This value is unknown to the graduate ex ante, but will ex post be: <br>
$$
    u_{i,j}^k = v_{j} + \epsilon_{i,j}^k
$$

They know that $\epsilon^k_{i,j}\sim \mathcal{N}(0,\sigma^2)$, but they do not observe $\epsilon^k_{i,j}$ before making their career choice. <br>

Consider the concrete case of $J=3$ with:
$$
\begin{align*}
    v_{1} &= 1 \\
    v_{2} &= 2 \\
    v_{3} &= 3
\end{align*}
$$

If the graduates know the values of $v_j$ and the distribution of $\epsilon_{i,j}^k$, they can calculate the expected utility of each career track using simulation: <br>
$$
    \mathbb{E}\left[ u^k_{i,j}\vert v_j \right] \approx v_j + \frac{1}{K}\sum_{k=1}^K \epsilon_{i,j}^k
$$

Now consider a new scenario: Imagine that the graduate does not know $v_j$. The *only* prior information they have on the value of each job, comes from their $F_{i}$ friends that work in each career $j$. After talking with them, they know the average utility of their friends (which includes their friends' noise term), giving them the prior expecation: <br>
$$
\tilde{u}^k_{i,j}\left( F_{i}\right) = \frac{1}{F_{i}}\sum_{f=1}^{F_{i}} \left(v_{j} + \epsilon^k_{f,j}\right), \; \epsilon^k_{f,j}\sim \mathcal{N}(0,\sigma^2)
$$
For ease of notation consider that each graduate have $F_{i}=i$ friends in each career. <br>

For $K$ times do the following: <br>
1. For each person $i$ draw $J\cdot F_i$ values of $\epsilon_{f,j}^{k}$, and calculate the prior expected utility of each career track, $\tilde{u}^k_{i,j}\left( F_{i}\right)$. <br>
Also draw their own $J$ noise terms, $\epsilon_{i,j}^k$
1. Each person $i$ chooses the career track with the highest expected utility: $$j_i^{k*}= \arg\max_{j\in{1,2\dots,J}}\left\{ \tilde{u}^k_{i,j}\left( F_{i}\right)\right\} $$
1. Store the chosen careers: $j_i^{k*}$, the prior expectation of the value of their chosen career: $\tilde{u}^k_{i,j=j_i^{k*}}\left( F_{i}\right)$, and the realized value of their chosen career track: $u^k_{i,j=j_i^{k*}}=v_{j=j_i^{k*}}+\epsilon_{i,j=j_i^{k*}}^k$.

Chosen values will be: <br>
$i\in\left\{1,2\dots,N\right\}, N=10$ <br>
$F_i = i$<br>
So there are 10 graduates. The first has 1 friend in each career, the second has 2 friends, ... the tenth has 10 friends.

After a year of working in their career, the graduates learn $u^k_{ij}$ for their chosen job $j_i^{k*}$ perfectly. <br>
The can switch to one of the two remaining careers, for which they have the same prior as before, but it will now include a switching cost of $c$ which is known.
Their new priors can be written as: 
$$
\tilde{u}^{k,2}_{ij}\left( F_{i}\right) = \begin{cases}
            \tilde{u}^k_{ij}\left( F_{i}\right)-c & \text{if } j \neq j_i^{k*} \\
            u_{ij=j_i^{k*}} & \text{if } j = j_i^{k*}
        \end{cases}
$$

We will set $c=1$.

Their realized utility will be: <br>
$$
u^{k,2}_{ij}= \begin{cases}
            u_{ij}^k -c & \text{if } j \neq j_i^{k*} \\
            u_{ij=j_i^{k*}} & \text{if } j = j_i^{k*}
        \end{cases}
$$

**Question 3:** Following the same approach as in question 2, find the new optimal career choice for each $i$, $k$. Then for each $i$, calculate the average subjective expected utility from their new optimal career choice, and the ex post realized utility of that career. Also, for each $i$, calculate the share of graduates that chooses to switch careers, conditional on which career they chose in the first year. <br>

```python
# Parameters
par = SimpleNamespace()
par.J = 3
par.N = 10
par.K = 10000  # number of simulations
par.sigma = 2
par.v = np.array([1, 2, 3])
par.c = 1

# Initialize storage for results
share_careers = np.zeros((par.N, par.J))
avg_subjective_utility = np.zeros(par.N)
avg_realized_utility = np.zeros(par.N)
share_switching = np.zeros(par.N)

# Initialize storage for conditional switching shares
conditional_switching = np.zeros((par.N, par.J))

# Simulation
for i in range(1, par.N + 1):  # For each graduate
    career_choices = np.zeros(par.J)
    subjective_utilities = []
    realized_utilities = []
    switch_count = 0
    initial_choice_counts = np.zeros(par.J)
    initial_to_switch_counts = np.zeros(par.J)
    
    for k in range(par.K):  # For each simulation
        # Generate noise for friends
        epsilon_friends = np.random.normal(0, par.sigma, (par.J, i))
        
        # Calculate prior expected utility
        prior_expected_utility = par.v + np.mean(epsilon_friends, axis=1)
        
        # Generate noise for the graduate
        epsilon_graduate = np.random.normal(0, par.sigma, par.J)
        
        # Choose career with highest expected utility
        initial_choice = np.argmax(prior_expected_utility)
        initial_choice_counts[initial_choice] += 1
        
        # Calculate subjective expected utility and realized utility
        initial_subjective_utility = prior_expected_utility[initial_choice]
        initial_realized_utility = par.v[initial_choice] + epsilon_graduate[initial_choice]
        
        # Calculate new priors for switching
        new_priors = np.where(
            np.arange(par.J) == initial_choice,
            initial_realized_utility,
            prior_expected_utility - par.c
        )
        
        # Determine new optimal career
        new_choice = np.argmax(new_priors)
        career_choices[new_choice] += 1
        
        # Store subjective expected utility and realized utility
        subjective_utility = new_priors[new_choice]
        realized_utility = (par.v[new_choice] + epsilon_graduate[new_choice]) if new_choice != initial_choice else initial_realized_utility
        
        subjective_utilities.append(subjective_utility)
        realized_utilities.append(realized_utility)
        
        # Check if switched and update counts
        if new_choice != initial_choice:
            switch_count += 1
            initial_to_switch_counts[initial_choice] += 1
    
    # Store results
    share_careers[i-1, :] = career_choices / par.K
    avg_subjective_utility[i-1] = np.mean(subjective_utilities)
    avg_realized_utility[i-1] = np.mean(realized_utilities)
    share_switching[i-1] = switch_count / par.K
    conditional_switching[i-1, :] = initial_to_switch_counts / np.maximum(initial_choice_counts, 1)  # avoid division by zero

# Visualization

# Share of graduates choosing each career
plt.figure(figsize=(12, 5))
for j in range(par.J):
    plt.plot(range(1, par.N + 1), share_careers[:, j], marker = 'o', label=f'Career {j + 1}')
plt.xlabel('Graduate')
plt.ylabel('Share Choosing Career')
plt.title('Share of Graduates Choosing Each Career After Considering Switching Costs')
plt.xticks(range(1, par.N + 1))
plt.legend()
plt.grid(True)
plt.show()

# Average subjective expected utility
plt.figure(figsize=(12, 5))
plt.bar(range(1, par.N + 1), avg_subjective_utility, color = "darkred", label='Subjective Expected Utility')
plt.xticks(range(1, par.N + 1))
plt.ylim(2.5,3.5)
plt.xlabel('Graduate')
plt.ylabel('Utility')
plt.title('Average Subjective Expected Utility After Switching Decision')
plt.legend()
plt.grid(True)
plt.show()

# Average realized utility
plt.figure(figsize=(12, 5))
plt.bar(range(1, par.N + 1), avg_realized_utility, color='darkgreen', label='Realized Utility')
plt.xticks(range(1, par.N + 1))
plt.ylim(2.5,3.5)
plt.xlabel('Graduate')
plt.ylabel('Utility')
plt.title('Average Realized Utility After Switching Decision')
plt.legend()
plt.grid(True)
plt.show()

# Share of graduates switching
#plt.figure(figsize=(12, 5))
#plt.bar(range(1, par.N + 1), share_switching, color='red', label='Share Switching')
#plt.xticks(range(1, par.N + 1))
#plt.xlabel('Graduate')
#plt.ylabel('Share Switching')
#plt.title('Share of Graduates Switching Careers After One Year')
#plt.legend()
#plt.grid(True)
#plt.show()

# Conditional switching shares
plt.figure(figsize=(12, 5))
for j in range(par.J):
    plt.bar(range(1, par.N + 1), conditional_switching[:, j], label=f'Initial Career {j + 1}')
plt.xlabel('Graduate')
plt.ylabel('Share Switching')
plt.xticks(range(1, par.N + 1))
plt.title('Share of Graduates Switching Careers Conditional on Initial Career')
plt.legend(bbox_to_anchor=(1.3, 1.05))
plt.grid(True)
plt.show()
```

```python
# Parameters
J = 3  # Number of careers
N = 10  # Number of graduates
K = 10000  # Number of simulations
sigma = 2  # Standard deviation of noise
v = np.array([1, 2, 3])  # True value of each career
c = 1  # Switching cost

# Initialize storage
chosen_careers = np.zeros((K, N), dtype=int)
subjective_utilities = np.zeros((K, N))
realized_utilities = np.zeros((K, N))
new_chosen_careers = np.zeros((K, N), dtype=int)
new_subjective_utilities = np.zeros((K, N))
new_realized_utilities = np.zeros((K, N))

# Simulation
for k in range(K):
    for i in range(1, N + 1):
        F_i = i  # Number of friends
        epsilon_friends = np.random.normal(0, sigma, (J, F_i))  # Friends' noise
        epsilon_personal = np.random.normal(0, sigma, J)  # Personal noise
        
        # Prior expected utility based on friends
        prior_utility = v + np.mean(epsilon_friends, axis=1)
        
        # Initial career choice
        chosen_career = np.argmax(prior_utility)
        
        # Store initial choices and utilities
        chosen_careers[k, i - 1] = chosen_career
        subjective_utilities[k, i - 1] = prior_utility[chosen_career]
        realized_utilities[k, i - 1] = v[chosen_career] + epsilon_personal[chosen_career]
        
        # Perfect knowledge after a year
        known_utility = realized_utilities[k, i - 1]
        
        # Calculate new priors with switching cost
        new_prior_utility = np.copy(prior_utility)
        new_prior_utility[chosen_career] = known_utility  # No switching cost for chosen career
        new_prior_utility[prior_utility != known_utility] -= c  # Apply switching cost
        
        # New career choice
        new_chosen_career = np.argmax(new_prior_utility)
        
        # Store new choices and utilities
        new_chosen_careers[k, i - 1] = new_chosen_career
        new_subjective_utilities[k, i - 1] = new_prior_utility[new_chosen_career]
        if new_chosen_career == chosen_career:
            new_realized_utilities[k, i - 1] = known_utility
        else:
            new_realized_utilities[k, i - 1] = v[new_chosen_career] + epsilon_personal[new_chosen_career] - c

# Calculate averages
average_new_subjective_utility = np.zeros(N)
average_new_realized_utility = np.zeros(N)
switch_share = np.zeros((N, J))

for i in range(N):
    average_new_subjective_utility[i] = np.mean(new_subjective_utilities[:, i])
    average_new_realized_utility[i] = np.mean(new_realized_utilities[:, i])
    for j in range(J):
        switch_share[i, j] = np.mean((chosen_careers[:, i] == j) & (new_chosen_careers[:, i] != j))

# Visualization
fig, ax = plt.subplots(1, 3, figsize=(18, 5))

# Bar plot for the share of graduates switching careers
sns.barplot(data=switch_share.T, ax=ax[0], errorbar = None)
ax[0].set_title("Share of Graduates Switching Careers")
ax[0].set_xlabel("Career Chosen in First Year")
ax[0].set_ylabel("Share Switching")
ax[0].grid(False)

# Line plot for average subjective expected utility
sns.lineplot(x=range(1, N+1), y=average_new_subjective_utility, marker='o', ax=ax[1])
ax[1].set_title("Average New Subjective Expected Utility")
ax[1].set_xlabel("Graduate Type")
ax[1].set_ylabel("Average Subjective Utility")

# Line plot for average realized utility
sns.lineplot(x=range(1, N+1), y=average_new_realized_utility, marker='o', color='green', ax=ax[2])
ax[2].set_title("Average New Ex Post Realized Utility")
ax[2].set_xlabel("Graduate Type")
ax[2].set_ylabel("Average Realized Utility")

plt.tight_layout()
plt.show()
```

## 3. <a id='toc3_'></a>[Problem 3: Barycentric interpolation](#toc0_)

**Problem:** We have a set of random points in the unit square,

$$
\mathcal{X} = \{(x_1,x_2)\,|\,x_1\sim\mathcal{U}(0,1),x_2\sim\mathcal{U}(0,1)\}.
$$

For these points, we know the value of some function $f(x_1,x_2)$,

$$
\mathcal{F} = \{f(x_1,x_2) \,|\, (x_1,x_2) \in \mathcal{X}\}.
$$

Now we want to approximate the value $f(y_1,y_2)$ for some  $y=(y_1,y_2)$, where $y_1\sim\mathcal{U}(0,1)$ and $y_2\sim\mathcal{U}(0,1)$.

**Building block I**

For an arbitrary triangle $ABC$ and a point $y$, define the so-called barycentric coordinates as:

$$
\begin{align*}
  r^{ABC}_1 &= \frac{(B_2-C_2)(y_1-C_1) + (C_1-B_1)(y_2-C_2)}{(B_2-C_2)(A_1-C_1) + (C_1-B_1)(A_2-C_2)} \\
  r^{ABC}_2 &= \frac{(C_2-A_2)(y_1-C_1) + (A_1-C_1)(y_2-C_2)}{(B_2-C_2)(A_1-C_1) + (C_1-B_1)(A_2-C_2)} \\
  r^{ABC}_3 &= 1 - r_1 - r_2.
\end{align*}
$$

If $r^{ABC}_1 \in [0,1]$, $r^{ABC}_2 \in [0,1]$, and $r^{ABC}_3 \in [0,1]$, then the point is inside the triangle.

We always have $y = r^{ABC}_1 A + r^{ABC}_2 B + r^{ABC}_3 C$.

B&=\arg\min_{(x_{1},x_{2})\in\mathcal{X}}\sqrt{\left(x_{1}-y_{1}\right)^{2}+\left(x_{2}-y_{2}\right)^{2}}\text{ s.t. }x_{1}>y_{1}\text{ and }x_{2}<y_{2}\\
C&=\arg\min_{(x_{1},x_{2})\in\mathcal{X}}\sqrt{\left(x_{1}-y_{1}\right)^{2}+\left(x_{2}-y_{2}\right)^{2}}\text{ s.t. }x_{1}<y_{1}\text{ and }x_{2}<y_{2}\\
D&=\arg\min_{(x_{1},x_{2})\in\mathcal{X}}\sqrt{\left(x_{1}-y_{1}\right)^{2}+\left(x_{2}-y_{2}\right)^{2}}\text{ s.t. }x_{1}<y_{1}\text{ and }x_{2}>y_{2}.
\end{align*}
$$
